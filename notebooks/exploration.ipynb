```python
movies = pd.read_csv("../data/tmdb_5000_movies.csv")

credits = pd.read_csv("../data/tmdb_5000_credits.csv")

print(movies.shape)

print(credits.shape)

movies.head()
```

```python
movies = movies.merge(credits, on = 'title')
movies.shape
```

```python
movies.columns
```

```python
movies = movies[['movie_id', 'title_x', 'overview', 'genres', 'keywords', 'cast_x', 'crew_x']]
```

```python
movies.rename(columns={
    'title_x': 'title',
    'cast_x': 'cast',
    'crew_x': 'crew'
}, inplace=True)
```

```python
movies.head()
```

```python
import ast

def convert(obj):
    L = []
    for i in ast.literal_eval(obj):
        L.append(i['name'])
    return L
```

```python
convert(movies.iloc[0].genres)
```

```python
movies['genres'] = movies['genres'].apply(convert)
movies['keywords'] = movies['keywords'].apply(convert)
```

```python
movies.head()
```

```python
def convert_cast(obj):
    L = []
    counter = 0
    for i in ast.literal_eval(obj):
        if counter < 3:
            L.append(i['name'])
        counter += 1
    return L
```

```python
convert_cast(movies.iloc[0].cast)
```

```python
movies['cast'] = movies['cast'].apply(convert_cast)
```

```python
movies.head()
```

```python
def fetch_director(obj):
    L = []
    for i in ast.literal_eval(obj):
        if i['job'] == 'Director':
            L.append(i['name'])
    return L
```

```python
fetch_director(movies.iloc[0].crew)
```

```python
movies['crew'] = movies['crew'].apply(fetch_director)
```

```python
movies.head()
```

```python
movies['tags'] = movies['overview'] + movies['genres'].apply(lambda x: ' '.join(x)) + movies['keywords'].apply(lambda x: ' '.join(x)) + movies['cast'].apply(lambda x: ' '.join(x)) + movies['crew'].apply(lambda x: ' '.join(x))
```

```python
movies[['title', 'tags']].head()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(max_features=5000, stop_words='english')
```

```python
movies['tags'] = movies['tags'].fillna('')
```

```python
vectors = tfidf.fit_transform(movies['tags'])
```

```python
vectors.shape
```

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity = cosine_similarity(vectors)
```

```python
similarity.shape
```

```python
def recommend(movie):
    movie_index = movies[movies['title'] == movie].index[0]
    distances = similarity[movie_index]
    
    movies_list = sorted(
        list(enumerate(distances)),
        reverse=True,
        key=lambda x: x[1]
    )[1:6]
    
    for i in movies_list:
        print(movies.iloc[i[0]].title)
```

```python
recommend('Iron Man')
```

```python
import pickle

pickle.dump(movies, open('movies.pkl', 'wb'))
pickle.dump(similarity, open('similarity.pkl', 'wb'))
```
